```python
# --- Action Tools (Regular Functions) ---

def square(num):
    """Takes a number and returns its square."""
    return num * num

def cube(num):
    """Takes a number and returns its cube."""
    return num * num * num


# --- The Higher-Order Function ---

def operate(num, operation):
    """
    Accepts a number AND a function reference ('operation').
    It applies the function to the number and returns the result.
    """
    # 'operation' acts as a placeholder for whatever function is passed in
    return operation(num)


# --- Executing the Code ---

my_number = 5

# Example A: Passing 'square' as an argument
# Notice we write 'square', NOT 'square()' because we are passing the tool itself, not its result.
result_1 = operate(my_number, square)
print("Square result:", result_1)  # Output: 25

# Example B: Passing 'cube' as an argument
result_2 = operate(my_number, cube)
print("Cube result:", result_2)    # Output: 125

```

```python

#############Processing Collections (Lists)

def square(num):
    return num * num

def cube(num):
    return num * num * num

# --- The Higher-Order Function for Collections ---

def process_list(numbers, action_rule):
    """
    Loops through a list and applies a dynamic 'action_rule'
    function to every single element.
    """
    for item in numbers:
        # Apply the passed-in function rule to the current list item
        calculated_value = action_rule(item)
        print(f"Processed value: {calculated_value}")


# --- Executing the Code ---

my_list = [5, 6, 7]

print("--- Running Square Action ---")
# Passing the 'square' rule to the list processor
process_list(my_list, square)

print("\n--- Running Cube Action ---")
# Passing the 'cube' rule to the list processor
process_list(my_list, cube)
```
